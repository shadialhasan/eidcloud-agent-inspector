# 🔍 EidCloud Agent Inspector - Interactive Quickstart

Welcome to **EidCloud Agent Inspector** ("Wireshark for AI Agents").
This notebook demonstrates how to load an execution trace, step through packet frames, filter events, and identify latency & token bottlenecks in pure PHP.

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/shadialhasan/eidcloud-agent-inspector/blob/main/notebooks/quickstart.ipynb)

```python
# Cell 1: Environment Setup & Clone
!apt-get update -qq && apt-get install -y -qq php-cli
!if [ ! -d "eidcloud-agent-inspector" ]; then git clone https://github.com/shadialhasan/eidcloud-agent-inspector.git; fi
%cd eidcloud-agent-inspector
!php -v
!php bin/eidcloud-inspector --help
```

```python
# Cell 2: Inspecting Session & Timeline
# Parse sample trace, display packet timeline and filter by specific tool
!php bin/eidcloud-inspector inspect examples/session_sample.json

print("\n--- Filtering for 'filesystem' tool calls ---")
!php bin/eidcloud-inspector filter examples/session_sample.json --tool=filesystem
```

```python
# Cell 3: Bottleneck Detection & Fast Replay
# Identify high latency or token spikes, then execute a 0x instant session replay
!php bin/eidcloud-inspector bottlenecks examples/session_sample.json --min-latency=1500 --min-tokens=2000

print("\n--- Running Turn-by-Turn Session Replay ---")
!php bin/eidcloud-inspector replay examples/session_sample.json --speed=0x
```
